### extract hidden data from file Ex3_sound1.wav ###
data is hidden in last significant bit of every byte  

#### Get data from the file ####
* import library wave, it will be used to extract bytes from audio file
* load the file into soundWithMsg variable using wave.open module function
* read frames to load sound bytes from audio file to variable extractedBytes
* close the file - release resources the moment we are done using it

```python
import wave  # used methods from https://docs.python.org/3/library/wave.html
soundWithMsg = wave.open("Ex3_sound1.wav", mode='rb') 
# extract all the bytes
extractedBytes = soundWithMsg.readframes(-1)  # data is in bytes object - literal format b'\xfe\xff\...', -1 is to read all frames 
soundWithMsg.close() #close the resources as soon as we are done using it 
```

#### Extract last bit from every byte and create new bytes #### 
* change from hex literal format of bytes b'\xfe\xff\...' to ints [254, 255, ...]
* change from int [254, 255, ...] to binary ['11111110','11111111', ...]
* extract last bit (last digit) in binary and form new list element every 8 bit extracted forming 1 byte of hidden message part ['01001101',
 '01101111', ...] 

```python
# change from literal format to list of integers - took the idea  from https://docs.python.org/3/library/stdtypes.html#bytes
extractedIntList = list(extractedBytes) # data is in a list of integer formal [254, 255, ...]

# change from int to binary 
extractedBinaryList = []
for i in range (len(extractedIntList)):
    extractedBinaryList.append(format(extractedIntList[i], 'b')) # will format from int to binary and append that to list ['11111110','11111111', ...]
    
# extract last bit from every byte which at this step is last binary digit 
lastBitBinaryList = []
for i in range (0,len(extractedBinaryList), 8): # iterate over Binary list and step every 8
    joinedLastBits = ""
    for j in range (0,8): # in every step of 8, take last digit from each binary
        joinedLastBits = joinedLastBits + extractedBinaryList[i + j][-1] # create string that is created by combingig last digit from 8 consecutive binary list items 
    lastBitBinaryList.append(joinedLastBits) # append that new string to the list 
```

#### Turn extracted binary hidden message to readable text ####
* change each binary list element ['01001101','01101111', ...] to int element [77, 111, ...]
* from int [77, 111, ...]  to unicode character 'Mo...'
* print extracted hidden message 

```python

# turn lastBitBinaryList to lastBitIntList as a path to later change from int to unicode readable characters
lastBitIntList = []
for i in range (len(lastBitBinaryList)):
    lastBitIntList.append(int(lastBitBinaryList[i], 2)) # change each binary element to int 
    
# from int to ASCII character 
decodedMsg = ""
for i in range (len(lastBitIntList)):
    if chr(lastBitIntList[i]) == '#': # after msg the rest would be filled with '#' so might as well stop at that point 
        break
    decodedMsg = decodedMsg + chr(lastBitIntList[i])
print(decodedMsg)
```
